```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Create a Sample Movie Dataset
data = {
    'Movie_ID': [1, 2, 3, 4, 5, 6, 7],
    'Title': [
        'The Dark Knight',
        'Inception',
        'Interstellar',
        'The Avengers',
        'Shutter Island',
        'Man of Steel',
        'The Matrix'
    ],
    'Genre_Description': [
        'action superhero crime batman fighting dark city savior',
        'sci-fi thriller dreams subconscious mind heist reality inception',
        'sci-fi space exploration wormhole time travel gravity universe interstellar',
        'action superhero marvel aliens saving the world earth team',
        'psychological thriller mystery mental hospital crime investigation shutter island',
        'action superhero superman alien saving world flying superpower',
        'sci-fi action virtual reality simulation machines hacking matrix'
    ]
}

df = pd.DataFrame(data)

# Step 2: TF-IDF Vectorization
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['Genre_Description'])

# Step 3: Calculate Cosine Similarity Matrix
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# Step 4: Recommendation Function
def get_movie_recommendations(title, cosine_sim=cosine_sim, df=df, top_n=3):
    # Check if movie exists in dataset
    if title not in df['Title'].values:
        return f"Movie '{title}' not found in database."
    
    # Get index of the target movie
    idx = df[df['Title'] == title].index[0]
    
    # Pair similarity score with index
    sim_scores = list(enumerate(cosine_sim[idx]))
    
    # Sort movies based on similarity score (descending)
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    
    # Get top N most similar movies (excluding the movie itself at index 0)
    sim_scores = sim_scores[1:top_n+1]
    
    # Retrieve movie indices
    movie_indices = [i[0] for i in sim_scores]
    
    # Format results
    results = pd.DataFrame({
        'Recommended Movie': df['Title'].iloc[movie_indices].values,
        'Similarity Score': [round(i[1], 4) for i in sim_scores]
    })
    
    return results

# Step 5: Test the System
user_input = "Inception"
print(f"--- Top Recommendations for '{user_input}' ---")
print(get_movie_recommendations(user_input, top_n=3))
```
